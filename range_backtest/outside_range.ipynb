```python
import pandas as pd
import numpy as np
df=pd.read_json(r"D:\FM\candle_maker\artifacts\hist_data.json")
```

```python
df.shape
```

```python
df = df[df["date"] >= "2026-05-06"]
df = df[df["symbol"].isin(["BPCL","NTPC","RELIANCE","SBIN","VEDL","POWERGRID","HINDALCO","INFY"])]
```

```python
df.shape
```

```python
df['next_day_open'] = df.groupby('symbol')['open'].shift(-1)
df['next_day_date'] = df.groupby('symbol')['date'].shift(-1)
df['signal'] = np.where((df['next_day_open'] > df['high'])| (df['next_day_open'] < df['low']), 1, 0)
```

```python
df=df[df['signal']==1]
```

```python
df=df[['date','next_day_date','symbol','signal']]
```

```python
df.sort_values(by='date',inplace=True)
df.to_csv('outside_range.csv',index=False)
```
